# Update the Edition Profile

I have decided to change the edition profile so that each witness, source, etc., will have the following fields:

- human_label: a human-readable short-form reference that could be displayed on screen or in print.
- machine_label: a variation on human_label that does not include punctuation or spaces.
- id: a UUID.
- type: a type from a controlled vocabulary (e.g., "manuscript", "early_edition", "modern_edition", etc.)

There will also be fields for bibliographical information where needed.

```python
import json
import lxml.etree as ET
import pandas as pd
```

```python
profile = json.load(
    open("../json-copy/phi0830.phi001.edition_profile.json", "r", encoding="utf-8")
)
```

```python
tree = ET.parse("../xml/editio.xml")

# Register a namespace for the TEI elements
ns = {"tei": "http://www.tei-c.org/ns/1.0", "xml": "http://www.w3.org/XML/1998/namespace"}
```

```python
editio_xml_ids = set()
for xmlid in tree.xpath("//tei:witness/@xml:id", namespaces=ns):
    editio_xml_ids.add(xmlid)
for xmlid in tree.xpath("//tei:bibl/@xml:id", namespaces=ns):
    editio_xml_ids.add(xmlid)
for person in tree.xpath("//tei:person/@xml:id", namespaces=ns):
    editio_xml_ids.add(person)
```

```python
editio_abbrs = set()
for abbr in tree.xpath("//tei:witness/tei:abbr", namespaces=ns):
    abbr = abbr.xpath("string(.)").strip()
    editio_abbrs.add(abbr)
for abbr in tree.xpath("//tei:bibl/tei:abbr", namespaces=ns):
    abbr = abbr.xpath("string(.)").strip()
    editio_abbrs.add(abbr)
for abbr in tree.xpath("//tei:person/tei:abbr", namespaces=ns):
    abbr = abbr.xpath("string(.)").strip()
    editio_abbrs.add(abbr)
```

```python
wit_dict = {}
for wit in tree.xpath("//tei:witness", namespaces=ns):
    wit_id = wit.get("{http://www.w3.org/XML/1998/namespace}id")
    if wit_id is None or wit_id == "":
        wit_id = None
    wit_abbr = wit.xpath("string(tei:abbr)", namespaces=ns)
    if wit_abbr == "":
        wit_abbr = None
    wit_dict[wit_id] = wit_abbr
for bibl in tree.xpath("//tei:bibl", namespaces=ns):
    bibl_id = bibl.get("{http://www.w3.org/XML/1998/namespace}id")
    if bibl_id is None or bibl_id == "":
        bibl_id = None
    bibl_abbr = bibl.xpath("string(tei:abbr)", namespaces=ns)
    if bibl_abbr == "":
        bibl_abbr = None
    wit_dict[bibl_id] = bibl_abbr
for person in tree.xpath("//tei:person", namespaces=ns):
    person_id = person.get("{http://www.w3.org/XML/1998/namespace}id")
    if person_id is None or person_id == "":
        person_id = None
    person_abbr = person.xpath("string(tei:persName/tei:abbr)", namespaces=ns)
    if person_abbr == "":
        person_abbr = None
    wit_dict[person_id] = person_abbr
```

```python
for index, object in enumerate(profile["sigla"]):
    if object["siglum"] in wit_dict:
        object["label"] = wit_dict[object["siglum"]]
        print(f"Updated label for {object['siglum']}: {object['label']}")
        object["id"] = object["siglum"].replace(" ", "").replace(".", "")
        print(f"Updated id for {object['siglum']}: {object['id']}")
    else:
        print(f"No matching siglum found for {object['siglum']}")
```
