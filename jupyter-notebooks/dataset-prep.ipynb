# Prepare the Dataset

I want to split the sample data into two parts: development and evaluation. This is not a typical machine-learning experiment, so I don't need training, validation, test/evaluation sets. I just need a small sample to see if the prompts are returning good results, and an evaluation set for the full run. Both sets need to represent the full range of phenomena in the entries and claims.

````python
import json
````

````python
with open('../sample/gold-standard.json', 'r') as f:
    data = json.load(f)
````

````python
len(data)
````

There are 74 objects in the array of entries. But there is one that spans a page break, which I have decided to handle manually, so I should remove it. It is the entry for line 1.13.

````python
# Find the index for the object with key: "target": ["urn:cts:latinLit:phi0830.phi001.dll:1.13@vocas"]
for i, obj in enumerate(data):
    if obj.get("target") == ["urn:cts:latinLit:phi0830.phi001.dll:1.13@vocas"]:
        del data[i]
        print(f"Deleted at index {i}: {obj}")
        break
````

I should aim for about 30% of the data as the development set.

````python
len(data) * .30
````

I'll round up and see whether the first 22 entries are representative of the data.

````python
dev = data[0:22]
````

````python
len(dev)
````

````python
print(json.dumps(dev, indent=4, ensure_ascii=False))
````

That's a good sample, but it leaves out entries for 1.10, 1.11. and 1.12 on the first page. If I include those, I can just use the first page as a test set.

````python
dev = data[0:28]
````

````python
print(json.dumps(dev, indent=4, ensure_ascii=False))
````

````python
dev[27]
````

````python
evaluation = data[28:]
print(json.dumps(evaluation, indent=4, ensure_ascii=False))
````

Now I need to make two JSON files from `dev` and `evaluation`. They should be in the form:

```json
{
    "split": "splitname",
    "entry_ids": {
        "uuid1",
        "uuid2"
        "…",
    }
}
```


````python
dev_object = {
    "split": "development",
    "entry_ids": [obj["id"] for obj in dev]
}
json.dump(dev_object, open('../sample/dev.json', 'w'), indent=4, ensure_ascii=False)

evaluation_object = {
    "split": "evaluation",
    "entry_ids": [obj["id"] for obj in evaluation]
}
json.dump(evaluation_object, open('../sample/evaluation.json', 'w'), indent=4, ensure_ascii=False)
````
