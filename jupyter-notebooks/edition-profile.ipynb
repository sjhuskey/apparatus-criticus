```python
import json
```

```python
file = "../json/phi0830.phi001.editionprofile.json"
```

```python
data = json.load(open(file))
```

```python
data
```

```python
sigla = data["sigla"]
```

```python
new_sigla = {}
for siglum in sigla:
    for key, value in siglum.items():
        siglum = key
        name = value
        type = "manuscript"
        new_sigla[siglum] = {"siglum": siglum, "name": name, "type": type}
```

```python
new_sigla
```

```python
print(json.dumps(new_sigla, indent=4, ensure_ascii=False))
```

```python
import lxml.etree as ET
```

```python
tree = ET.parse("../xml/editio.xml")
```

```python
ns = {
    "tei": "http://www.tei-c.org/ns/1.0",
    "xml": "http://www.w3.org/XML/1998/namespace"
}

# Get all witness elements in the tei:listWit with xml:id="editions-witnesses"
witnesses = tree.xpath('//tei:listWit[@xml:id="editions-witnesses"]//tei:witness', namespaces=ns)
```

```python
for witness in witnesses:
    # List the elements in each witness
    siglum = witness.attrib.get("xml:id")
    print(siglum)
```
