# Extract `<abbr>` values from TEI XML

I need to extract the values of `abbr` elements to use in the edition profile.

```python
import lxml.etree as ET
import csv
```

```python
# Parse the XML file and create an ElementTree object
tree = ET.parse("../xml/editio.xml")

# Register a namespace for the TEI elements
ns = {"tei": "http://www.tei-c.org/ns/1.0", "xml": "http://www.w3.org/XML/1998/namespace"}
```

```python
def normalize_space(value: str | None) -> str:
    """Equivalent to XPath normalize-space()."""
    return " ".join((value or "").split())
```

```python
# Write the witness information to a CSV file
with open("CSV/witnesses.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Short Form", "Label", "Title"])

    for witness in tree.xpath("//tei:div[@type='bibliography']/tei:listWit/tei:witness", namespaces=ns):
        label_list = witness.xpath("tei:abbr[@type='siglum']/text()", namespaces=ns)
        title_list = witness.xpath("tei:title/text()", namespaces=ns)
        label = label_list[0] if label_list else None
        # Get the short form of the witness, which is the value of the xml:id attribute
        short_form = witness.get(f"{{{ns['xml']}}}id")
        title = normalize_space(title_list[0]) if title_list else None
        writer.writerow([short_form, label, title])
        print(f"Label: {label}, Title: {title}, Short Form: {short_form}")

    for source in tree.xpath(
        "//tei:div[@type='bibliography']/tei:listBibl/tei:bibl", namespaces=ns
    ):
        label_list = source.xpath("tei:abbr[@type='siglum']/text()", namespaces=ns)
        title_list = source.xpath("tei:title/text()", namespaces=ns)
        label = label_list[0] if label_list else None
        title = normalize_space(title_list[0]) if title_list else None
        short_form = source.get(f"{{{ns['xml']}}}id")
        writer.writerow([short_form, label, title])
        print(f"Label: {label}, Title: {title}, Short Form: {short_form}")
```
