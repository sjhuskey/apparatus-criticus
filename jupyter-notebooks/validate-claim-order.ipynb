# Validate Claim Order

I want to be sure that the claims in an apparatus entry are in the correct order:

1. Lemma (if present)
2. Reporting rank (1, 2, 3, etc.)

```python
import json
```

```python
with open('../sample/phi0830.phi001.dll:1.1-1.26.json', 'r') as f:
    data = json.load(f)
```

```python
for entry in data:
    claims = entry.get("claims", [])
    for i in range(len(claims) - 1):
        current_claim = claims[i]
        next_claim = claims[i + 1]
        if "reporting_rank" in current_claim and "reporting_rank" in next_claim:
            if (
                current_claim["reporting_rank"] is not None
                and next_claim["reporting_rank"] is not None
                and current_claim["reporting_rank"] >= next_claim["reporting_rank"]
            ):
                print(
                    f"Claim order issue in entry {entry.get('id', 'unknown')}: "
                    f"Claim {current_claim['id']} should come before Claim {next_claim['id']}."
                )
```

```python
for entry in data:
    claims = entry.get("claims", [])
    for i in range(len(claims) - 1):
        reading_type_current = claims[i].get("reading_type")
        if reading_type_current == "lemma" and i != 0:
            print(
                f"Claim order issue in entry {entry.get('id', 'unknown')}: "
                f"Claim {claims[i]['id']} with reading_type 'lemma' should be the first claim."
            )
```
