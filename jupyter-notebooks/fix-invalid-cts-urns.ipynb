# Fix Invalid CTS URNs

Some of Giarratano's apparatus entries refer to more than one word. I need to fix the CTS URN values for those instances that contain words with spaces. Instead of `urn:cts:latinLit:phi0830.phi001.dll:1.5@molle sub`, I want `urn:cts:latinLit:phi0830.phi001.dll:1.5@molle-sub`.

```python
import json
```

```python
with open(
    "../converted-json/merged-apparatus-entries-and-claims-updated-nested-corrected.json",
    "r",
    encoding="utf-8",
) as f:
    data = json.load(f)
```

```python
def fix_invalid_urn(urn):
    if urn is None:
        raise ValueError("URN is None")
    if not isinstance(urn, str):
        raise ValueError("URN must be a string")

    if "@" not in urn:
        return urn

    if urn.count("@") > 1:
        raise ValueError(f"Invalid URN: {urn}")

    base, substring = urn.split("@")
    words = substring.split()

    if len(words) <= 1:
        return urn

    return f"{base}@{words[0]}-{words[-1]}"
```

```python
# Fix the text_locus field for each entry in the data
for entry in data:
    print(f"Processing entry with id: {entry.get('id')}")
    # Fix the text_locus field if it is a list
    if isinstance(entry.get("text_locus"), list):
        entry["text_locus"] = [fix_invalid_urn(urn) for urn in entry["text_locus"]]
    else:
        # If text_locus is not a list, fix it directly
        entry["text_locus"] = fix_invalid_urn(entry.get("text_locus"))
```

```python
# Fix the target field for each claim in the data
for entry in data:
    if entry.get("type") == "claim":
        target = entry.get("target")
        if isinstance(target, list):
            entry["target"] = [fix_invalid_urn(urn) for urn in target]
        else:
            entry["target"] = fix_invalid_urn(target)
```
