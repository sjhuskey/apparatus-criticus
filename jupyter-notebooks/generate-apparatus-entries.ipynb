# Generate Apparatus Entries from Processed CSV

I need to create JSON with `"type": "apparatus_enty"` for all the values in the `entries_with_cts.csv` that I created in the Jupyter notebook `markdown-to-xml.json`.

```python
import json
import pandas as pd
from datetime import datetime, timezone
```

```python
df = pd.read_csv("../converted-json/entries_with_cts5.csv")
```

```python
df
```

```python
id_cts = {}
with open("../converted-json/calpSicPoem1-target-array.json", "r") as f:
    data = json.load(f)
    # Loop through the array of objects and create a dictionary with id as key and cts as value
    for item in data:
        # If the apparatus_entry_id is empty, skip this item
        if "apparatus_entry_id" not in item or not item["apparatus_entry_id"]:
            continue
        else:
            id_cts["".join(item["target"])] = item["apparatus_entry_id"]
```

```python
id_cts
```

```python
def create_json_object(row):
    if row['validate'] is True and row['cts_id'] in id_cts:
        return {
            "type": "apparatus_entry",
            "id": id_cts[row['cts_id']],
            "target": [row['cts_id']],
            "source_edition": "urn:uuid:062e759d-619b-4549-b50b-e73f421609e3",
            "source_page": None,
            "source_image": None,
            "bbox_norm": [],
            "transcription": row['xml'],
            "transcription_confidence": "high",
            "transcriber": "SJH",
            "extraction": {
                "model": None,
                "extracted_at": datetime.now(timezone.utc).strftime(
                    "%Y-%m-%dT%H:%M:%S.%fZ"
                ),
                "extraction_image": None,
            },
        }
    else:
        return None
```

```python
entries = []
for index, row in df.iterrows():
    entry = create_json_object(row)
    if entry is not None:
        entries.append(entry)
    else:
        print(f"Row {index} is invalid or cts_id not found in id_cts.")
```

```python
entries
```

```python
with open("../converted-json/apparatus_entries.json", "w") as f:
    json.dump(entries, f, indent=4, ensure_ascii=False)
```

```python
def create_all_json_object(row):
    return {
        "type": "apparatus_entry",
        "id": [],
        "target": [row["cts_id"]],
        "source_edition": "urn:uuid:062e759d-619b-4549-b50b-e73f421609e3",
        "source_page": None,
        "source_image": None,
        "bbox_norm": [],
        "transcription": row["xml"],
        "transcription_confidence": "high",
        "transcriber": "SJH",
        "extraction": {
            "model": None,
            "extracted_at": datetime.now(timezone.utc).strftime(
                "%Y-%m-%dT%H:%M:%S.%fZ"
            ),
            "extraction_image": None,
        },
    }
```

```python
all_entries = []
for index, row in df.iterrows():
    entry = create_all_json_object(row)
    if entry is not None:
        all_entries.append(entry)
    else:
        print(f"Row {index} is invalid or cts_id not found in id_cts.")
```

```python
for entry in all_entries:
    print(json.dumps(entry, indent=4, ensure_ascii=False))
```
