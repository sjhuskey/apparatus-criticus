```python
import json
from pathlib import Path
```

```python
def add_reporting_rank(json_dir):
    json_dir_path = Path(json_dir)

    for json_file in json_dir_path.glob("*.json"):
        with json_file.open("r+") as f:
            data = json.load(f)

            reporting_rank = 1

            for index, obj in enumerate(data):
                if (
                    obj.get("type") == "apparatus_claim"
                    and obj.get("claim_type") == "reading"
                ):
                    new_obj = {}

                    for key, value in obj.items():
                        if key == "reporting_rank":
                            continue

                        new_obj[key] = value

                        if key == "supporting_evidence":
                            new_obj["reporting_rank"] = reporting_rank

                    data[index] = new_obj
                    reporting_rank += 1

            with json_file.open("w", encoding="utf-8") as f:
                json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
json_dir = "../json_copy"
add_reporting_rank(json_dir)
```
