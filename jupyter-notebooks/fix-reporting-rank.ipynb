# Fix Reporting Rank

I wasn't consistent in adding `reporting_rank` to claims. I need to make sure that every claim has one.

```python
import json
```

```python
with open('../sample/phi0830.phi001.dll:1.1-1.26.json') as f:
    data = json.load(f)
```

```python
for entry in data:
    for claim in entry['claims']:
        # If there isn't a reporting_rank, report it.
        if 'reporting_rank' not in claim:
            print(f"Missing reporting_rank for claim {claim['id']}")
```

```python
# Make sure that reporting ranks are in order and unique within each apparatus entry
for entry in data:
    ranks = [claim['reporting_rank'] for claim in entry['claims'] if 'reporting_rank' in claim]
    if len(ranks) != len(set(ranks)):
        print(f"Duplicate reporting ranks found in apparatus entry {entry['id']}")
    if sorted(rank for rank in ranks if rank is not None) != list(range(1, len(ranks) + 1)):
        print(f"Reporting ranks are not sequential in apparatus entry {entry['id']}")
```

```python
# Make sure that every claim array has a reporting_rank that starts at 1 if there is more than one claim.
for entry in data:
    ranks = [
        claim["reporting_rank"]
        for claim in entry["claims"]
        if "reporting_rank" in claim
    ]
    if (
        len([rank for rank in ranks]) >= 1
        and min([int(rank) for rank in ranks if rank is not None]) != 1
    ):
        print(f"Reporting ranks do not start at 1 in apparatus entry {entry['id']}")
```

```python
# If there is a lemma claim, make sure it has reporting_rank 1 and all other claims have reporting_rank > 1.
for entry in data:
    claims = entry["claims"]

    if any(c["reading_type"] == "lemma" for c in claims):
        for claim in claims:
            if claim["reading_type"] == "lemma":
                claim["reporting_rank"] = 1
            else:
                claim["reporting_rank"] += 1
```

```python
with open('../sample/gold-standard.json', 'w') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
```
