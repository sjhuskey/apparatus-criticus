# Ensure that reported lemmata match the target

I need to make sure that any reading reported as a `lemma` matches the target expressed in the CTS URN.

```python
import json
```

```python
with open('../sample/phi0830.phi001.dll:1.1-1.26.json') as f:
    data = json.load(f)
```

```python
for entry in data:
    for claim in entry['claims']:
        if claim['reading_type'] == 'lemma':
            target = claim['target'][0]
            if '@' in target:
                urn, lemma = target.split('@')
                claim['urn'] = urn
                claim['lemma'] = lemma
            if claim['reading'] != claim['lemma']:
                print(f"Mismatch: {claim['reading']} != {claim['lemma']}")
```
