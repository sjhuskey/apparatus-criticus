```python
import json
import pandas as pd
from uuid import uuid4
```

```python
json_file = json.load(open("../json/phi0830.phi001.editionprofile.json", encoding="utf-8"))
```

```python
for siglum in json_file['sigla']:
    # Give each siglum a new UUID
    siglum['uuid'] = str(uuid4())
```

```python
for abbreviation in json_file['secondary_sources']:
    # Give each abbreviation a new UUID
    abbreviation['uuid'] = str(uuid4())
```

```python
for person in json_file['people']:
    # Give each person a new UUID
    person['uuid'] = str(uuid4())
```

```python
# Save the updated JSON file with UUIDs, keep unicode characters intact
with open("../json/phi0830.phi001.editionprofile_with_uuids_new.json", "w", encoding="utf-8") as f:
    json.dump(json_file, f, ensure_ascii=False, indent=4)
```

```python
old_json_file = json.load(open("../json/phi0830.phi001.editionprofile_with_uuids.json", encoding="utf-8"))
new_json_file = json.load(open("../json/phi0830.phi001.editionprofile_with_uuids_new.json", encoding="utf-8"))
```

```python
# If the old_json_file does not have a uuid in the secondary_sources, add it from the new_json_file
for abbreviation in old_json_file['secondary_sources']:
    if 'uuid' not in abbreviation:
        new_abbreviation = next((abbrev for abbrev in new_json_file['secondary_sources'] if abbrev['abbreviation'] == abbreviation['abbreviation']), None)
        if new_abbreviation:
            abbreviation['uuid'] = new_abbreviation['uuid']
```

```python
with open(
    "../json/phi0830.phi001.editionprofile_with_uuids_final.json", "w", encoding="utf-8"
) as f:
    json.dump(old_json_file, f, ensure_ascii=False, indent=4)
```
