# Validate UUID+Label Pairs

I need to make sure that only one ID is associated with each UUID and vice versa.

```python
import json
from collections import defaultdict
```

```python
with open('../sample/phi0830.phi001.dll:1.1-1.26.json', 'r') as f:
    data = json.load(f)
```

```python
id_label_map = defaultdict(str)
```

```python
for entry in data:
    for claim in entry.get('claims', []):
        for evidence in claim.get('supporting_evidence', []):
            if 'id' in evidence and 'label' in evidence:
                if 'id' in id_label_map:
                    if id_label_map[evidence['id']] != evidence['label']:
                        print(f"Conflict for ID {evidence['id']}: {id_label_map[evidence['id']]} vs {evidence['label']}")
                else:
                    id_label_map[evidence['id']] = evidence['label']
```

```python
label_id_map = defaultdict(str)
for entry in data:
    for claim in entry.get('claims', []):
        for evidence in claim.get('supporting_evidence', []):
            if 'id' in evidence and 'label' in evidence:
                if 'label' in label_id_map:
                    if label_id_map[evidence['label']] != evidence['id']:
                        print(f"Conflict for label {evidence['label']}: {label_id_map[evidence['label']]} vs {evidence['id']}")
                else:
                    label_id_map[evidence['label']] = evidence['id']
# Sort the label_id_map by label
sorted_label_id_map = dict(sorted(label_id_map.items()))
sorted_label_id_map
```
