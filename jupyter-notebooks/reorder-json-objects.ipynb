# Reorder JSON objects

Although I am relating `apparatus_claims` to their `apparatus_entry` by the `apparatus_entry_id` key, it would make more sense to nest those claims in the `apparatus_entry`. I'll do that here.

````python
import json
````

````python
file = "../converted-json/merged-apparatus-entries-and-claims-updated.json"

with open(file, "r", encoding="utf-8") as f:
    data = json.load(f)
````

````python
def reorder_keys(entry):
    if entry.get("type") == "apparatus_entry":
        ordered_entry = {
            "type": entry.get("type"),
            "id": entry.get("id"),
            "text_locus": entry.get("text_locus"),
            "source_edition": entry.get("source_edition"),
            "source_page": entry.get("source_page"),
            "source_image": entry.get("source_image"),
            "bbox_norm": entry.get("bbox_norm"),
            "transcription": entry.get("transcription"),
            "transcriber": entry.get("transcriber"),
            "extraction": entry.get("extraction"),
        }
        return ordered_entry


def get_apparatus_claims(entry, data=data):
    claims = []
    for claim in data:
        if claim.get("type") == "apparatus_claim" and claim.get(
            "apparatus_entry_id"
        ) == entry.get("id"):
            claims.append(claim)
    return claims
````

````python
new_data = []
for entry in data:
    if entry.get("type") == "apparatus_entry":
        ordered_entry = reorder_keys(entry)
        claims = get_apparatus_claims(entry, data)
        ordered_entry["claims"] = claims
        new_data.append(ordered_entry)
    elif entry.get("type") == "apparatus_claim":
        continue  # Skip claims, as they are nested under entries
    
````
