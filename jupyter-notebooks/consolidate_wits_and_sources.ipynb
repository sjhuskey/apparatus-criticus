```python
import json
from pathlib import Path
```

```python
def combine_witnesses_and_sources(json_dir):
    
    json_dir_path = Path(json_dir)
    
    for json_file in json_dir_path.glob("*.json"):
        with json_file.open("r") as f:
            data = json.load(f)
            for index, obj in enumerate(data):
                if obj.get("type") == "apparatus_claim":
                    new_obj = {}

                    for key, value in obj.items():

                        new_obj[key] = value

                        if key == "supporting_evidence":
                            new_value = []
                            # If there are witnesses objects, add them to the supporting_evidence list
                            if "witnesses" in value:
                                new_value.extend(value["witnesses"])
                            if "sources" in value:
                                new_value.extend(value["sources"])
                            new_obj[key] = new_value
                    data[index] = new_obj

            with json_file.open("w", encoding="utf-8") as f:
                json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
json_dir = "../json_copy"
combine_witnesses_and_sources(json_dir)
```
